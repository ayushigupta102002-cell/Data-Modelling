Q. List all the places you want to travel someday.

```python
places = ["Goa", "Andaman and Nicobar Islands", "Mansarovar Lake", "Trincomalee", "Cambridge University", "Switzerland","Abisko","Great Barrier Reef","Sahara Desert","Salar de Uyuni"]
print("Top ten places I want to visit are:")
for place in places:
    print(place)
```

```python
print(len(places) == 10)
sum(places.count(place)for place in places) == len(places)
```

```python
places1 = places.copy()
```

```python
places1.insert(places1.index("Trincomalee"),"Kedarnath")
print("Before any international trips, I want to visit Kedarnath.")
places1.remove("Cambridge University")
print("Cambridge University is only for academic purposes and hence removed.")
places1.append("Antarctica")
print("I want to visit Antarctica as well.")
places1.pop(0)
print("I have already visited Goa hence removed it from the list.")
```

```python
print("Post final revision, top ten places I want to visit are:")
for place in places1:
    print(place)
```

```python
print(len(places1) == 10)
# approach 2 to check for duplicates in the list. more efficent than the previous approach.
len(set(places1)) == len(places1) 
```

```python
destinations = places1.copy()
```

```python
print("type of places is:", type(places1))
print("type of destinations is:", type(destinations))
for place in destinations:
    print(type(place))
```

```python
print("First destination:", destinations[0])
print("Second destination:", destinations[1])
print("Third destination:", destinations[2])
print("Last destination:", destinations[-1])
print("Second to last destination:", destinations[-2])
print("Third to last destination:", destinations[-3])
print("Remaining destinations:", destinations[3:-3])
```

```python
print("first 3 destinations are all within india:")
for x in destinations[:3]:
    print(x)

print("")

print("while other destinations are all international:")
for x in destinations[3:]:
    print(x)
```
