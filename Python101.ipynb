```python
x = "Ayushi Gupta"; print("The type of", x, " is:", type(x))
x = "5"; print("The type of", x, " is:", type(x))
x = 5; print("The type of", x, " is:", type(x))
x = 5.0; print("The type of", x, " is:", type(x))
x = 0j + 5; print("The type of", x, " is:", type(x))
x = [5]; print("The type of", x, " is:", type(x))
x = (5,5); print("The type of", x, " is:", type(x))
x = range(5); print("The type of", x, " is:", type(x))
x = {"integer" : 5}; print("The type of", x, " is:", type(x))
x = {"integer","string", "float", "complex"}; print("The type of", x, " is:", type(x))
x = frozenset({"integer","string", "float", "complex"}); print("The type of", x, " is:", type(x))
x = True; print("The type of", x, " is:", type(x))
x = b"Hello"; print("The type of", x, " is:", type(x))
x = bytearray(5); print("The type of", x, " is:", type(x))
x = memoryview(bytes(5)); print("The type of", x, " is:", type(x))
x = None; print("The type of", x, " is:", type(x))
```

```python
# Create a function for factorial
def factorial(n):
    # Return the factorial of n
    if n <= 1:
      print(1)
    else:
      x = n
      initial = n
      while x > 1:
        initial *=x-1
        x = x-1
        final = initial
      return final
```

```python
print(factorial(0))
```
